```python
hotel_summary = (
    df.groupby('hotel')
      .agg(
          bookings=('hotel', 'size'),
          cancellations=('is_canceled', 'sum'),
          avg_adr=('adr', 'mean'),
          total_revenue=('estimated_revenue', 'sum'),
          avg_stay_nights=('total_stay_nights', 'mean'),
          avg_lead_time=('lead_time', 'mean')
      )
      .reset_index()
)

hotel_summary['cancellation_rate'] = (
    hotel_summary['cancellations'] / hotel_summary['bookings'] * 100
)

hotel_summary['avg_revenue_per_booking'] = (
    hotel_summary['total_revenue'] / hotel_summary['bookings']
)

hotel_summary = hotel_summary.round({
    'avg_adr': 2,
    'total_revenue': 2,
    'avg_stay_nights': 2,
    'avg_lead_time': 2,
    'cancellation_rate': 2,
    'avg_revenue_per_booking': 2
})

hotel_summary.to_csv(base_path + 'hotel_summary.csv', index=False)

hotel_summary
```

```python
from sklearn.ensemble import IsolationForest

anomaly_features = [
    'adr',
    'total_stay_nights',
    'lead_time',
    'total_guests',
    'total_of_special_requests'
]

X = df[anomaly_features].copy()

iso_forest = IsolationForest(
    contamination=0.02,
    random_state=42
)

df['anomaly_flag'] = iso_forest.fit_predict(X)

df['anomaly_flag'] = df['anomaly_flag'].map({
    -1: 1,
     1: 0
})

anomaly_summary = (
    df.groupby(['hotel', 'anomaly_flag'])
      .agg(
          booking_count=('hotel', 'size'),
          avg_adr=('adr', 'mean'),
          avg_stay_nights=('total_stay_nights', 'mean'),
          avg_lead_time=('lead_time', 'mean'),
          avg_guests=('total_guests', 'mean'),
          avg_special_requests=('total_of_special_requests', 'mean'),
          estimated_revenue=('estimated_revenue', 'sum')
      )
      .reset_index()
)

hotel_totals = (
    df.groupby('hotel')
      .size()
      .reset_index(name='hotel_total_bookings')
)

anomaly_summary = anomaly_summary.merge(
    hotel_totals,
    on='hotel',
    how='left'
)

anomaly_summary['percentage_of_hotel_bookings'] = (
    anomaly_summary['booking_count']
    / anomaly_summary['hotel_total_bookings']
    * 100
)

anomaly_summary['status'] = anomaly_summary['anomaly_flag'].map({
    1: 'Anomaly',
    0: 'Normal'
})

anomaly_summary = anomaly_summary[
    [
        'hotel',
        'status',
        'booking_count',
        'percentage_of_hotel_bookings',
        'avg_adr',
        'avg_stay_nights',
        'avg_lead_time',
        'avg_guests',
        'avg_special_requests',
        'estimated_revenue'
    ]
]

anomaly_summary = anomaly_summary.round(2)

anomaly_summary.to_csv(base_path + 'anomaly_summary.csv', index=False)

print("Anomaly summary created successfully.")
print(f"Total anomalies: {(df['anomaly_flag'] == 1).sum()}")
print(f"Total normal bookings: {(df['anomaly_flag'] == 0).sum()}")

anomaly_summary
```
