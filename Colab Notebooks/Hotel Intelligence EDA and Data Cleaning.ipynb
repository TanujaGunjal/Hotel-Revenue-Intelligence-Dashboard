**Data Cleaning & Data Quality**

```python
df_raw = df.copy()
```

```python
df["children"] = df["children"].fillna(0)
df["country"] = df["country"].fillna("Unknown")
df["agent"] = df["agent"].fillna(0)
df["company"] = df["company"].fillna(0)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate rows:", len(duplicate_rows))
print("Duplicate groups:", df.duplicated().sum())
```

```python
duplicate_analysis = (
    duplicate_rows
    .groupby(["hotel", "is_canceled"])
    .size()
    .reset_index(name="count")
)

duplicate_analysis
```

```python
duplicate_rows["reservation_status"].value_counts()
```

```python
before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

```python
df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d"
)
```

```python
df[[
    "arrival_date_year",
    "arrival_date_month",
    "arrival_date_day_of_month",
    "arrival_date"
]].head()
```

```python
df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)
```

```python
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)
```

```python
df["estimated_revenue"] = (
    df["adr"] * df["total_stay_nights"]
)
```

```python
print("Negative ADR:", (df["adr"] < 0).sum())
print("Zero ADR:", (df["adr"] == 0).sum())

print("Zero adults:", (df["adults"] == 0).sum())

print("Zero guests:", (df["total_guests"] == 0).sum())

print("Zero-night stays:", (df["total_stay_nights"] == 0).sum())
```

```python
df[df["adr"] < 0][[
    "hotel",
    "is_canceled",
    "adults",
    "children",
    "babies",
    "total_stay_nights",
    "adr",
    "reservation_status"
]]
```

```python
df[df["adults"] == 0][[
    "hotel",
    "is_canceled",
    "adults",
    "children",
    "babies",
    "total_stay_nights",
    "adr",
    "reservation_status"
]].head(20)
```

```python
df["adr_invalid"] = df["adr"] < 0

print("Invalid ADR records:", df["adr_invalid"].sum())
```

```python
df.loc[df["adr_invalid"], "adr"] = np.nan
```

```python
zero_adults = df[df["adults"] == 0]

zero_adults[[
    "hotel",
    "is_canceled",
    "children",
    "babies",
    "total_guests",
    "total_stay_nights",
    "adr",
    "reservation_status"
]].head(20)
```

```python
df["invalid_guest_count"] = df["total_guests"] == 0

print(
    "Records with zero total guests:",
    df["invalid_guest_count"].sum()
)
```

```python
Q1 = df["adr"].quantile(0.25)
Q3 = df["adr"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
df["adr_outlier"] = (
    (df["adr"] < lower_bound) |
    (df["adr"] > upper_bound)
)

print("ADR outliers:", df["adr_outlier"].sum())
```

```python
for col in ["meal", "market_segment", "distribution_channel"]:
    print(
        col,
        "Undefined:",
        (df[col] == "Undefined").sum()
    )
```

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Negative ADR",
        "Zero Adult Records",
        "Zero Guest Records",
        "ADR Outliers"
    ],
    "Count": [
        len(df),
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        (df["adr"] < 0).sum(),
        (df["adults"] == 0).sum(),
        (df["total_guests"] == 0).sum(),
        df["adr_outlier"].sum()
    ]
})

quality_report
```
